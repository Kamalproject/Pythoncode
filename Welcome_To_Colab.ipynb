```
# Write a program in python to Find the sum of digits of a given number.

num=int(input("Enetr a number: "))
sum=0
while(num>0):
  rem=num%10
  sum=sum+rem
  num=num//10

print("Sum of the digits of this number is ", sum)
```

```
#Write a program in python to Reverse a number.

num=int(input("Enter a number: "))
reverse=0
temp=num
while(temp>0):
  rem=temp%10
  reverse=reverse*10+rem
  temp//=10
print("Reverse of number ", num," is ", reverse)
```

```
#Write a program in python to Check whether a number is prime or not.

number=int(input("Enter a number: "))
count=1
for i in range(1,number):
  if(number%i==0):
    count+=1
if(count==2):
  print("Numebr ",number," is a Prime number")
else:
  print("Number ",number," is not a prime number")
```

```
# Write a program in python to Check a number is Armstrong or not.

number=int(input("Enter a number: "))
temp=number
sum=0
n=len(str(number))
while(temp>0):
  rem=temp%10
  sum=sum+rem**n
  temp//=10

if(sum==number):
  print("Number ",number," is Aramstrong number")
else:
  print("Number", number," is not Aramstrong number")
```

```
# Write a program in python to Check whether a number is palindrome or not.

number=int(input("Enter a number: "))
temp=number
reverse=0

while(temp!=0):
  rem=temp%10
  reverse=reverse*10+rem
  temp//=10
if(number==reverse):
  print("Number ",number," is Palindrome ")
else:
  print("Numbe ",number," is not palindrome")
```

```
#Write a program in python to Check whether a number is perfect or not.

number=int(input("Enter a number: "))
sum=0;
for i in range(1,number):
  if(number%i==0):
    sum+=i
if(number==sum):
  print("Number ", number," is Perfect number")
else:
  print(f"Number {number} is not a perfect number")
```

```
# Write a program in python to generate a Fibonacci series.

a=int(input("Enter a number: "))
b=int(input("Enter another number: "))

print(f"Fibonacci series of number {a},{b} is ")
for i in range (0,5):
  c=a+b
  a=b
  b=c
  print(c,end=" ")
```
